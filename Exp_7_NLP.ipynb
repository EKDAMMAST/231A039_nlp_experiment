```python
import spacy

# Load the spaCy English model
nlp = spacy.load("en_core_web_sm")

# Sample text
text = "Mr. Rizvi of Rizvi College is in India. India is a country in South Asia. It is one of the oldest civilizations."

# Process the text
doc = nlp(text)
print(doc)

# Extract named entities
for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```

```python
# Import Required Libraries
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser
```

```python
# Sample sentence
sentence = "The quick brown fox jumps over the lazy dog."
sentence1 = "This is the class of natural language processing."
```

```python
# Tokenization and POS tagging
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
tokens = word_tokenize(sentence1)
tagged = pos_tag(tokens)

tagged
```

```python
# Step 3: Create a chunk parser and parse the sentence
grammar = r"""
  NP: {<DT>?<JJ>*<NN>}
"""
chunk_parser = RegexpParser(grammar)
chunked = chunk_parser.parse(tagged)

# Display the chunked output
print(chunked)
```

```python
from nltk.tree import Tree
chunked.pretty_print()
```

```python
print("Name : Atharv Bhandare / Uin No. : 231A039")
text = "In this world where all humans and non-human toons coexist, is Dimen City, a bustling metropolis where people from all walks of life can be met. One such is Penny McCartney, a sharp-tongued, quick-witted legal arms dealer and gunslinger navigating her life along with her best friend Walter Sterling, a tech genius who keeps things logical, and their shared robot companion, 8."

tokens = word_tokenize(text)
print(tokens)
```

```python
print("Name : Atharv Bhandare / Uin No. : 231A039")
named_entities = ne_chunk(tagged_tokens)
print(named_entities)
```

```python
print("Name : Atharv Bhandare / Uin No. : 231A039")
for subtree in named_entities:
  if isinstance(subtree, nltk.Tree):
    entity = "".join([word for word, tag in subtree.leaves()])
    label = subtree.label()
    print(f"Entity: {entity}, Label: {label}")
```
