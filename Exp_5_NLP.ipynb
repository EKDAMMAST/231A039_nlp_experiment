```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
import nltk
from nltk import word_tokenize, bigrams, trigrams, FreqDist, ConditionalFreqDist
from nltk.corpus import stopwords
from collections import defaultdict

nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
def preprocess(text):
  text = text.lower()
  tokens = word_tokenize(text)
  stopwords_eng = set(stopwords.words('english'))
  tokens = [token for token in tokens if token.isalnum() and token not in stopwords_eng]
  return tokens
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
corpus = 'This is Alex. He enjoys playing cricket. This is Sarah. She likes reading books with Alex. Rahul is not interested in their activities.'
corpus
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
tokens = preprocess(corpus)
tokens
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
bigrams_list = list(bigrams(tokens))
bigrams_list
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
bigram_frequency = FreqDist(bigrams_list)
bigram_frequency
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
conditional_frequency = ConditionalFreqDist(bigrams_list)
conditional_frequency.items()
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
trigrams_list = list(trigrams(tokens))
trigrams_list
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
def predict_next_word(word):
  word = word.lower()
  if word in conditional_frequency:
    next_word = conditional_frequency[word].max()
    return next_word
  else:
    return None
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
def generate_text(start_word, num_words=5):
  text = start_word
  current_word = start_word
  for i in range(num_words):
    next_word = predict_next_word(current_word)
    if next_word is None:
      break
    text +=  " " + next_word
    current_word = next_word
  return text
```

```python
print("Name : Dhanush Pallan / Uin No. : 231A038")
generate_text("personal")
```
